# Pemroses Transkrip Nilai Akademik

```python

# ============================================
# PROGRAM : Pemroses Transkrip Nilai Akademik
# MATKUL  : Pengenalan Pemrograman
# ============================================

data_mahasiswa = input("Masukkan data mahasiswa (Nama, NIM, Tugas, UTS, UAS): ")

nama, nim, nilai_tugas, nilai_uts, nilai_uas = data_mahasiswa.split(",")

nama = nama.strip()
nim = nim.strip()
nilai_tugas = float(nilai_tugas.strip())
nilai_uts = float(nilai_uts.strip())
nilai_uas = float(nilai_uas.strip())

bobot_tugas = 0.20
bobot_uts = 0.35
bobot_uas = 0.45

nilai_akhir = (
    (nilai_tugas * bobot_tugas) + (nilai_uts * bobot_uts) + (nilai_uas * bobot_uas)
)

bobot_total = bobot_tugas + bobot_uts + bobot_uas

print("\n+" + "-" * 48 + "+")
print(f"|{'TRANSKRIP NILAI AKADEMIK':^48}|")
print("+" + "-" * 48 + "+")

print(f"| {'Nama Mahasiswa':<22} | {nama:<21} |")
print(f"| {'NIM':<22} | {nim:<21} |")

print("+" + "-" * 48 + "+")

print(f"| {'Komponen Nilai':<22} | {'Nilai':>9} | {'Bobot':>9} |")
print("+" + "-" * 48 + "+")

print(f"| {'Nilai Tugas':<22} | {nilai_tugas:>9.1f} | {bobot_tugas:>8.0%}  |")
print(f"| {'Nilai UTS':<22} | {nilai_uts:>9.1f} | {bobot_uts:>8.0%}  |")
print(f"| {'Nilai UAS':<22} | {nilai_uas:>9.1f} | {bobot_uas:>8.0%}  |")

print("+" + "-" * 48 + "+")

print(f"| {'NILAI AKHIR':<22} | {nilai_akhir:>9.1f} | {bobot_total:>8.0%}  |")

print("+" + "-" * 48 + "+")
```
